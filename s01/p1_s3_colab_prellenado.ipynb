## Parte A - Con conocimiento vs sin conocimiento

Imagina que le preguntas algo a un agente sobre una **politica interna de Linko**
que solo esta en un documento de la empresa.

- **Sin conocimiento:** el agente adivina con lo que sabe del mundo. Puede inventar.
- **Con conocimiento:** el agente lee el documento y responde con la fuente correcta.

Abajo tenemos un "documento" de ejemplo (unas pocas frases) y una pregunta.


### A.1 El documento y la pregunta

En la celda de abajo puedes **cambiar la pregunta** (donde dice `CAMBIA AQUI`).
Deja el documento como esta la primera vez; luego, si quieres, cambialo tambien.


```python
# El "documento" de la empresa (unas frases de ejemplo)
documento = """
Politica de viaticos Linko: el limite diario de alimentos es de 800 pesos.
Los taxis se reembolsan solo con factura. El hospedaje requiere autorizacion
previa del lider de area. Los viaticos se solicitan con 3 dias de anticipacion.
"""

# CAMBIA AQUI: escribe tu pregunta sobre el documento
pregunta = "Cual es el limite diario para alimentos?"

print("Pregunta:", pregunta)
```

### A.2 Respuesta SIN conocimiento

El agente responde sin ver el documento. Ejecuta y observa: puede inventar o dar
algo generico.

```python
# Sin conocimiento (resultado guardado de ejemplo)
respuesta_sin = ("No tengo informacion especifica sobre la politica de viaticos "
    "de tu empresa. En general, los limites de alimentos varian por organizacion; "
    "te sugiero consultar el manual interno correspondiente.")
print("SIN conocimiento:\n")
print(respuesta_sin)
```

### A.3 Respuesta CON conocimiento

Ahora le damos el documento junto con la pregunta. Ejecuta y compara: responde con
el dato correcto y basado en la fuente.

```python
# Con conocimiento (resultado guardado de ejemplo)
respuesta_con = ("Segun la politica de viaticos de Linko, el limite diario para "
    "alimentos es de 800 pesos.")
print("CON conocimiento:\n")
print(respuesta_con)
```

> **Lo importante:** la misma pregunta, dos respuestas muy distintas. Con el
> documento, el agente da el dato correcto (800 pesos) y no inventa. Eso es lo
> que lograste al darle conocimiento a tu agente.

## Parte B - Busqueda por significado

Cuando buscas "auto" y el sistema tambien encuentra "coche" y "vehiculo", eso es
**busqueda por significado**: entiende que las palabras se parecen aunque no sean
iguales.

Abajo tienes una lista de frases y una frase de busqueda. La IA ordenara las
frases de la mas parecida en significado a la menos parecida.


### B.1 Las frases y la busqueda

Puedes **cambiar la frase de busqueda** y, si quieres, las frases de la lista.
Ejecuta y mira como se ordenan por parecido de significado.

```python
# Lista de frases (puedes cambiarlas si quieres)
frases = [
    "El cliente quiere cancelar su contrato",
    "Necesito dar de baja mi servicio",
    "Cuando llega mi pedido?",
    "Estado de mi envio, por favor",
    "Quiero un reembolso",
]

# CAMBIA AQUI: escribe la frase que quieres buscar
busqueda = "Deseo terminar mi suscripcion"

print("Busco algo parecido a:", busqueda)
```

### B.2 Resultado ordenado por significado

Ejecuta. La IA pone primero las frases mas parecidas en significado a tu busqueda,
aunque usen palabras distintas.

```python
# Resultado guardado de ejemplo (para la busqueda de arriba)
resultados = [
    (0.86, "Necesito dar de baja mi servicio"),
    (0.83, "El cliente quiere cancelar su contrato"),
    (0.61, "Quiero un reembolso"),
    (0.42, "Estado de mi envio, por favor"),
    (0.39, "Cuando llega mi pedido?"),
]
print("De mas parecido a menos parecido:\n")
for sim, f in resultados:
    print(f"  {sim:.2f}  ->  {f}")
```

> **Lo importante:** "Deseo terminar mi suscripcion" no comparte casi ninguna
> palabra con "Necesito dar de baja mi servicio", pero la IA entiende que
> significan casi lo mismo. Por eso tu agente encuentra la respuesta correcta en
> tus documentos aunque tu pregunta use otras palabras.
